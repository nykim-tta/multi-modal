텍스트 LLM 으로 각 설명 문장에서 claim 을 추출하고 임베딩 합의를 계산해 **claim 집합을 파일로 저장**합니다.

- 비교 대상 LLM · 임베딩: `safety_pipeline/config.py` 의 `CLAIM_MODELS`, `EMBEDDING_MODELS`
- 산출물: `comparison_results/stage_a/stageA_claims__<LLM>__<EMB>.json` → Stage B 의 입력


```python
%load_ext autoreload
%autoreload 2

from safety_pipeline import config
from safety_pipeline.dataio import load_samples_from_data_dir
from safety_pipeline.stage_a import run_stage_a
```

```python
samples = load_samples_from_data_dir(config.DEFAULT_DATA_DIR)
print(f"로드된 샘플 수: {len(samples)}")
for s in samples:
    print(" -", s["sample_id"], s["image_path"])
```

```python
# LLM(×임베딩) 조합별로 claim 집합 생성 · 저장
claim_sets = run_stage_a(
    samples,
    experiment_config=config.EXPERIMENT_CONFIG,
    output_dir=config.DEFAULT_STAGE_A_DIR,
)
for (llm, emb), cs_list in claim_sets.items():
    print(f"[{llm} × {emb}] 샘플 {len(cs_list)}개, 총 claim {sum(len(cs.claims) for cs in cs_list)}개")
```
